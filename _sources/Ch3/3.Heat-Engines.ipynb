## Rankine Cycle

```python
Image(os.path.join(path, "Rankine Cycle.png"))
```

A Rankine cycle is the dominant thermodynamic cycle used for large‑scale electricity generation in the United States. Coal, biomass, concentrated solar thermal energy, and nuclear heat sources all drive a high‑pressure working fluid (typically steam) through:

1. A **pump**  
2. A **boiler**  
3. A **turbine**  
4. A **condenser**

The fluid is then recycled, forming a closed loop.


### Pathways of the Rankine Cycle

| Step | State | Unit Operation | Path | Heat | Work |
|:---:|:-----:|:--------------:|:-----|:-----|:-----|
| **1** | Low‑pressure liquid | — | — | — | — |
| **1–2** | | Pump | Isentropic | $\dot{Q}=0$ | $\dot{W} = \dot{N}\,\underline{V}_1 (P_2 - P_1)$ |
| **2** | High‑pressure liquid | — | — | — | — |
| **2–3** | | Boiler | Isobaric | $\dot{Q} = \dot{N}(\underline{H}_3 - \underline{H}_2)$ | $\dot{W}=0$ |
| **3** | Superheated vapor | — | — | — | — |
| **3–4** | | Turbine | Isentropic | $\dot{Q}=0$ | $\dot{W} = \dot{N}(\underline{H}_3 - \underline{H}_4)$ |
| **4** | Low‑quality vapor | — | — | — | — |
| **4–1** | | Condenser | Isobaric | $\dot{Q} = \dot{N}(\underline{H}_4 - \underline{H}_1)$ | $\dot{W}=0$ |


### Figure of Merit: Rankine‑Cycle Efficiency

The thermal efficiency of any heat engine is:

$$
\eta = -\frac{\dot{W}}{\dot{Q}_H}
$$

For the Rankine cycle:

$$
\eta 
= \frac{\dot{W}_{1-2} + \dot{W}_{3-4}}{\dot{Q}_{2-3}}
= \frac{(\underline{H}_4 - \underline{H}_3) - \underline{V}_1(P_2 - P_1)}{\underline{H}_3 - \underline{H}_2}
$$

> **Reversible Devices Assumed**  
> The efficiency expressions above assume *ideal*, *reversible* operation of both the pump and the turbine.  
> Real Rankine cycles experience significant irreversibilities, especially in the turbine, which reduce actual efficiency.

### Carnot Benchmark

Modern boilers:  
- $P_{\max} \approx 30\ \text{MPa}$  
- $T_{\max} \approx 650^\circ\text{C}$

Condensation occurs below $100^\circ\text{C}$.

$$
\eta_C = 1 - \frac{T_C}{T_H}
= 1 - \frac{373.15}{923.15}
= 59.6\%
$$

Real Rankine efficiencies are much lower due to irreversibilities, moisture limits, and material constraints.

#### Example Rankine Cycle (Using Simplifying Assumptions)

Assumptions:

1. Water enters boiler at $100^\circ\text{C}$, $30\ \text{MPa}$  
2. Leaves boiler at $650^\circ\text{C}$  
3. Steam exits turbine at $0.1\ \text{MPa}$

State table:

| State | $T$ (°C) | $P$ (MPa) | $\underline{H}$ (kJ/mol) | $\underline{S}$ (J/mol·K) |
|:----:|:----------:|:-----------:|:---------------------------:|:----------------------------:|
| 1 | 100 | 0.1 | 48.19 | 24.5 |
| 2 | 100 | 30 | 7.95 | 24.5 |
| 3 | 650 | 30 | 64.84 | 115.43 |
| 4 | 100 | 0.1 | 42.08 | 115.43 |

Assume molar volume of liquid water:

$$
\underline{V}_1 \approx \underline{V}_2 = 0.0180\ \text{L/mol}
$$

#### Efficiency:

$$
\eta 
= \frac{(42.08 - 64.84) - 0.018(30 - 0.1)}{7.95 - 64.84}
= 21.5\%
$$

This illustrates why real Rankine cycles fall far below the Carnot limit.

## Brayton Cycle

### 1. Overview
The **Brayton cycle** describes the idealized thermodynamic cycle for **gas‑turbine engines**, including jet engines and many stationary power plants.  
It models a **continuous‑flow**, **open system** using an ideal gas (usually air) as the working fluid.

The ideal Brayton cycle consists of **four internally reversible processes**:

1. **Isentropic compression** (compressor)  
2. **Constant‑pressure heat addition** (combustor or external heater)  
3. **Isentropic expansion** (turbine) 
4. **Constant‑pressure heat rejection** (exhaust or external cooler)

### 2. Process Diagram

1 → 2 : Isentropic compression  
2 → 3 : Constant‑pressure heat addition  
3 → 4 : Isentropic expansion  
4 → 1 : Constant‑pressure heat rejection

### 3. Thermodynamic Processes

#### Process 1 → 2: Isentropic Compression (Compressor)
- Working fluid is compressed adiabatically and reversibly.  
- Temperature and pressure rise.  
- Entropy remains constant.  
- Work **input** is required.

$$
T_2 = T_1 \left(\frac{P_2}{P_1}\right)^{\frac{\gamma-1}{\gamma}}
$$


#### Process 2 → 3: Constant‑Pressure Heat Addition
- Heat is added at constant pressure.  
- Temperature increases significantly.  
- Models combustion in a real gas turbine.

$$
\dot{Q} = C_P^* (T_3 - T_2)
$$


#### Process 3 → 4: Isentropic Expansion (Turbine)
- High‑temperature gas expands through the turbine.  
- Produces **work output**.  
- Entropy remains constant.

$$
T_4 = T_3 \left(\frac{P_4}{P_3}\right)^{\frac{\gamma-1}{\gamma}}
$$

#### Process 4 → 1: Constant‑Pressure Heat Rejection
- Heat is rejected at constant pressure.  
- Temperature drops back to the initial state.

$$
\dot{Q} = C_P^* (T_4 - T_1)
$$

### 4. Net Work and Thermal Efficiency

$$
\dot{W}_{net} = \dot{W}_{turbine} - \dot{W}_{compressor}
$$

### 5. Thermal Efficiency (Ideal Brayton Cycle)
Expressed in terms of the **pressure ratio** $ r = \frac{P_2}{P_1}$:

$$
\eta_{th} = 1 - \frac{1}{r^{(\gamma-1)/\gamma}}
$$

Key insight:  
**Increasing the pressure ratio increases efficiency**, but only up to practical limits (material temperatures, compressor work, etc.).

#### Brayton Cycle in Tabular Form


| Step | State Description | Unit Operation | Path Type | Heat Term | Work Term |
|:---:|:------------------|:--------------|:----------|:-----------|:-----------|
| **(1)** | Low‑pressure gas (ambient) | — | — | — | — |
| **(1–2)** | | **Compressor** | **Isentropic compression** | $\dot{Q}=0$ | $\dot{W}_{1\to2} = \dot{N} C_p^*(T_2 - T_1)$ |
| **(2)** | High‑pressure gas | — | — | — | — |
| **(2–3)** | | **Combustor** | **Isobaric heat addition** | $\dot{Q}_H = \dot{N} C_p^*(T_3 - T_2)$ | $\dot{W}=0$ |
| **(3)** | High‑T, high‑P gas | — | — | — | — |
| **(3–4)** | | **Turbine** | **Isentropic expansion** | $\dot{Q}=0$ | $\dot{W}_{3\to4} = \dot{N} C_p^*(T_4 - T_3)$ |
| **(4)** | Low‑pressure exhaust gas | — | — | — | — |
| **(4–1)** | | **Exhaust / Cooler** | **Isobaric heat rejection** | $\dot{Q}_C = \dot{N} C_p^*(T_1 - T_4)$ | $\dot{W}=0$ |

### 6. Real‑World Limitations

Real gas turbines deviate from the ideal cycle due to:
- Compressor and turbine **irreversibilities**  
- **Pressure losses** in combustion chambers  
- **Finite temperature limits**  
- **Non‑ideal gas behavior** at high temperatures  

Enhancements include:
- **Regeneration** (recuperators)  
- **Intercooling**  
- **Reheat**  

These modifications reshape the $T‑$underline{S}$ diagram and improve efficiency.


### 7. Key Takeaways
- The Brayton cycle models **gas‑turbine engines** using an ideal gas.  
- Efficiency depends strongly on **pressure ratio**.  
- Two processes are isentropic; two are constant‑pressure.  
- Real engines require modifications to approach ideal performance.  

### 8. Deployment in the US

Natural‑gas Brayton‑cycle turbines are widely used as **peaker plants**, generators that run only during periods of **high electricity demand** (hot summer afternoons, cold snaps, grid contingencies).  
Their value comes from **fast start‑up**, **high ramp rates**, and **low capital cost** relative to baseload technologies.

Brayton‑cycle gas turbines are ideal for peaking service because they offer:

- **Rapid start‑up** (minutes instead of hours)  
- **High power‑to‑weight ratio**  
- **Simple cycle configuration** → fewer components, faster thermal response  
- **Fuel flexibility**, but natural gas dominates due to pipeline availability  
- **Lower fixed costs** since they operate infrequently  

Efficiency is lower than combined‑cycle plants, but **speed and flexibility** matter more for peaking.

- **Start time:** 5–15 minutes for many modern units  
- **Capacity factor:** Very low (often <10%)

- **Emissions:** Lower CO₂ per MWh than coal, but higher than combined‑cycle due to lower efficiency  
- **Dispatch:** Called upon during price spikes or reliability events  

Peaker turbines support:

- **Load following** during rapid demand increases  
- **Voltage and frequency stability**  
- **Backup for variable renewables** (wind/solar)  
- **Emergency reserve capacity**  

As renewable penetration grows, peakers increasingly act as **fast‑ramping grid stabilizers** rather than long‑duration generators.

- Extensive U.S. **pipeline infrastructure**  
- High energy density and clean combustion  
- Quick controllability  
- Lower capital cost than batteries (for short‑duration, high‑power events)  



## Piston Based Heat Engines

In heat engines that derive useful work from the action of a moving boundary (i.e., pistons), we should appreciate that at a thermodynamic cycle must be completed that turns the action of compression and expansion of the cylinder volume to changes in internal energy of the gas within that volume. Based on this understanding at each point throughout the cycle state of the gas will be determined determined by the relative amounts of heat added and removed from the cylinder, the pathway taken between each state point, and the volume change undertaken by the cylinder during expansion.

If the expansion and compression cycles are reversible, then a mass, energy, and entropy balance around the closed volume  at any step throughout the cycle reveals the following equation:

1. $dN=0$
2. $dU=\dot{Q}-PdV$
3. $dS=\dfrac{\dot{Q}}{T}$

Combining 1 and 3 by equating the heat terms, we can see that:

$\quad dA=dU-TdS=-PdV$

In other words, the free energy change associated with expansion and contraction is described by the **Helmholtz Free Energy**, which represents the energy available for to do useful work. Because of the sine of the right hand side of the equation above, we can appreciate that:

- if $dV>0$, then $dA<0$ and the free energy of the gas has been decreased. In other words, expansion reduces the capacity of the gas to do useful work

- if $dV<0$, then $dA>0$ and the free energ yof the gas has been increased. In other words, compression increases the capacity of the gas to do useful work.

The details of different closed system cycles will depend on the path used to effect a chance of state of the gas.

- reversible, adiabatic expansion/compression is always **isoentropic**, (i.e., $dS=0$)

- **isothermal** expansion/compression of an ideal gas is always at constant internal energy (i.e., $dU=0$)

- **isobaric** expansion/compression are those a the volume change is affected through the exchange of heat where the piston undergoes expansion/compression without exerting a net force on the surroundings.

## Carnot Cycle

The Carnot cycle is the fundamental theoretical model for a heat engine operating between two thermal reservoirs. It represents the **maximum possible efficiency** any heat engine can achieve when converting heat into work. Although no real engine operates exactly as a Carnot engine, the cycle provides a benchmark for evaluating real systems.

Consider the following setup:

- $N$ moles of an ideal gas in a piston–cylinder  
- Compression and expansion occur **reversibly**  
- The cylinder is perfectly insulated except when in contact with a thermal reservoir  
- The piston is connected to a shaft from which we extract work  
- The cylinder head can contact either a **hot reservoir** at $T_H$ or a **cold reservoir** at $T_C$

Based on these assumptions, the Carnot engine operates through the following **four reversible processes**.

### Pathways of the Carnot Cycle

| Step | Description | Pathway | Heat | Work |
|:---:|:-----------:|:--------|:-----|:-----|
| **1–2** | Cold reservoir | Isothermal compression at $T_C$ | $\dot{Q}_C$ (heat rejected) | $-PdV$ |
| **2–3** | Recovery stroke | Adiabatic (isentropic) compression | $0$ | $-PdV$ |
| **3–4** | Hot reservoir | Isothermal expansion at $T_H$ | $\dot{Q}_H$ (heat absorbed) | $-PdV$ |
| **4–1** | Power stroke | Adiabatic (isentropic) expansion | $0$ | $-PdV$ |

This sequence forms a closed loop in both the $P\!-\!V$ and $T\!-\!S$ planes.

### State Table for an Ideal‑Gas Carnot Cycle

Assume:

- Compression ratio $K = P_2/P_1$  
- Ideal‑gas behavior  
- Constant heat capacities  

| Step | Pathway | $T$ | $P$ | $\underline{V}$ | $\underline{S}$ | $\underline{U}$ |
|:---:|:--------|:-----:|:------:|:------------------:|:------------------:|:------------------:|
| **1** | Cold isotherm | $T_C$ | $P_1$ | $\underline{V}_1$ | $0$ | $0$ |
| **1–2** | Isothermal compression | | | | | |
| **2** | | $T_C$ | $P_2 = K P_1$ | $\underline{V}_2 = \frac{RT_C}{P_2}$ | $-R\ln K$ | $0$ |
| **2–3** | Isentropic compression | | | | | |
| **3** | | $T_H$ | $P_3 = P_2\left(\frac{T_H}{T_C}\right)^{C_p/R}$ | $\underline{V}_3 = \frac{RT_H}{P_3}$ | $-R\ln K$ | $C_v(T_H - T_C)$ |
| **3–4** | Isothermal expansion | | | | | |
| **4** | | $T_H$ | $P_4 = P_1\left(\frac{T_H}{T_C}\right)^{C_p/R}$ | $\underline{V}_4 = \frac{RT_H}{P_4}$ | $0$ | $C_v(T_H - T_C)$ |
| **4–1** | Isentropic expansion | | | | | |

### Specified Quantities

- Hot‑reservoir temperature: $T_H$  
- Cold‑reservoir temperature: $T_C$  
- Compression ratio: $K = P_2/P_1$  
- Initial pressure: $P_1$  
- Reference state: $T_R = T_C$, $\underline{S}_1 = 0$  
- Ideal‑gas heat capacity: $C_v^{IG}$ (constant)  

### Governing Relations

#### Ideal‑Gas Equation of State
$$
P\underline{V} = RT
$$

#### Entropy Change of an Ideal Gas
$$
d\underline{S} = \frac{C_v}{T}dT + \frac{R}{\underline{V}}d\underline{V}
$$
$$
d\underline{S} = \frac{C_p}{T}dT - \frac{R}{P}dP
$$

#### Isothermal Processes
$$
\Delta U = 0,\qquad Q = W = RT\ln\left(\frac{V_f}{V_i}\right)
$$

#### Isentropic Processes
$$
TV^{\gamma-1} = \text{const},\qquad PV^\gamma = \text{const}
$$
or equivalently:

$$\bigg(\frac{T_f}{T_i}\bigg)^{C_p/R} = \frac{P_f}{P_i} $$


### Efficiency of the Carnot Cycle

Because heat is added isothermally at $T_H$ and rejected isothermally at $T_C$:

$$
\eta = 1 - \frac{Q_C}{Q_H}
$$

For an ideal gas undergoing reversible isothermal processes:

$$d\underline{S} = \dfrac{C_V^*}{T}dT - \dfrac{R}{\underline{V}} d\underline{V}$$

Isothermal transformations:

$$\Delta\underline{S} = R \ln \dfrac{\underline{V}_2}{\underline{V}_1}$$

$$
Q_H = T\Delta \underline{S} = RT_H \ln \dfrac{P_2}{P_1},\qquad Q_C = RT_C \ln \dfrac{P_4}{P_3}
$$

or 

$$Q_H = R T_H \ln K,\qquad Q_C = R T_C \ln 1/K $$

because $P_4/P_3 = 1/K$, this cancels when the ration of the heat is calculated, thus:

$$
\boxed{\eta = 1 + Q_C/Q_H = 1 - \frac{T_C}{T_H}}
$$

This is the **maximum possible efficiency** for any heat engine operating between $T_H$ and $T_C$.



```python
import numpy as np
import matplotlib.pyplot as plt

R = 8.314  # J/mol/K

def Carnot_IG(Th=800, Tc=298, P1=1, Cv=28, K=10):
    """
    Carnot cycle for an ideal gas.

    Inputs:
        Th : hot reservoir temperature (K)
        Tc : cold reservoir temperature (K)
        P1 : minimum pressure (bar)
        Cv : constant-volume heat capacity (J/mol/K)
        K  : compression ratio (P2/P1)

    Outputs:
        V_, P_, T_, S_, U_, A_ : state-point arrays
        PVpath_, TSpath_       : pathway arrays for plotting
    """

    Cp = Cv + R
    gamma = Cp / Cv

    # -
    # STATE POINTS
    # -
    P1 = P1 * 1e5  # bar → Pa
    V1 = R * Tc / P1
    T1 = Tc
    S1 = 0
    U1 = 0
    A1 = U1 - T1 * S1

    # State 2: isothermal compression at Tc
    P2 = K * P1
    V2 = R * Tc / P2
    T2 = Tc
    S2 = -R * np.log(K)
    U2 = 0
    A2 = U2 - T2 * S2

    # State 3: isentropic compression to Th
    # Correct isentropic relation:
    # T2 * (V2/V3)^(gamma-1) = Th
    V3 = V2 * (T2 / Th) ** (1 / (gamma - 1))
    P3 = R * Th / V3
    T3 = Th
    S3 = S2
    U3 = Cv * (Th - Tc)
    A3 = U3 - T3 * S3

    # State 4: isothermal expansion at Th
    # Must return to V1 after isentropic expansion
    V4 = R * Th / (P1 * (Th / Tc) ** (gamma / (gamma - 1)))
    P4 = R * Th / V4
    T4 = Th
    S4 = 0
    U4 = Cv * (Th - Tc)
    A4 = U4 - T4 * S4

    # -
    # STATE ARRAYS (converted)
    # -
    V_ = np.array([V1, V2, V3, V4, V1]) * 1e6  # m³/mol → cm³/mol
    P_ = np.array([P1, P2, P3, P4, P1]) / 1e5  # Pa → bar
    T_ = np.array([T1, T2, T3, T4, T1]) - 273.15
    S_ = np.array([S1, S2, S3, S4, S1])
    U_ = np.array([U1, U2, U3, U4, U1])
    A_ = np.array([A1, A2, A3, A4, A1])

    # -
    # PV PATHWAYS
    # -
    # Isothermal at Tc
    V12 = np.logspace(np.log10(V1), np.log10(V2))
    P12 = R * Tc / V12

    # Isentropic to Th
    V23 = np.logspace(np.log10(V2), np.log10(V3))
    P23 = P2 * (V2 / V23) ** gamma

    # Isothermal at Th
    V34 = np.logspace(np.log10(V3), np.log10(V4))
    P34 = R * Th / V34

    # Isentropic back to Tc
    V41 = np.logspace(np.log10(V4), np.log10(V1))
    P41 = P4 * (V4 / V41) ** gamma

    PVpath_ = [
        [V12 * 1e6, P12 / 1e5],
        [V23 * 1e6, P23 / 1e5],
        [V34 * 1e6, P34 / 1e5],
        [V41 * 1e6, P41 / 1e5],
    ]

    # -
    # TS PATHWAYS (physically correct)
    # -
    # Isothermal: S changes, T constant
    S12 = np.linspace(S1, S2)
    T12 = np.full_like(S12, Tc)

    # Isentropic: S constant, T changes
    T23 = np.logspace(np.log10(T2), np.log10(T3))
    S23 = np.full_like(T23, S2)

    # Isothermal: S changes, T constant
    S34 = np.linspace(S3, S4)
    T34 = np.full_like(S34, Th)

    # Isentropic: S constant, T changes
    T41 = np.logspace(np.log10(T4), np.log10(T1))
    S41 = np.full_like(T41, S4)

    TSpath_ = [
        [T12 - 273.15, S12],
        [T23 - 273.15, S23],
        [T34 - 273.15, S34],
        [T41 - 273.15, S41],
    ]

    return V_, P_, T_, S_, U_, A_, PVpath_, TSpath_

V_,P_,T_,S_,U_,A_,PVpath_,TSpath_=Carnot_IG(Th=900,Cv=20,K=10)

labels = ["1", "2", "3", "4", "1"]

fig, ax = plt.subplots(dpi=300, figsize=(5,2), ncols=2)
fig.subplots_adjust(wspace=0.5)

colors = ['r','r','b','b']

# -
# PV plot with labels
# -
ax[0].plot(V_, P_, marker='s', linestyle='')

for i, path in enumerate(PVpath_):
    ax[0].plot(path[0], path[1], c=colors[i])

# Add labels near each state point
for i in range(4):
    ax[0].text(V_[i]*1.05, P_[i]*1.05, labels[i], fontsize=8)

ax[0].set_xscale('log')
ax[0].set_yscale('log')
ax[0].set_xlabel(r'$V$ (cm$^3$/mol)')
ax[0].set_ylabel(r'$P$ (bar)')


# -
# TS plot with labels
# -
ax[1].plot(S_, T_, marker='s', linestyle='')

for i, path in enumerate(TSpath_):
    ax[1].plot(path[1], path[0], c=colors[i])

# Add labels near each state point
for i in range(4):
    ax[1].text(S_[i]*1.02, T_[i]*1.02, labels[i], fontsize=8)

ax[1].set_xlabel(r'$S$ (J/mol/K)')
ax[1].set_ylabel(r'$T$ (°C)')
```

## Stirling Cycle

A Stirling cycle is a regenerative, closed‑cycle heat engine that uses an ideal gas as its working fluid. Unlike the Carnot cycle, which uses two isothermal and two adiabatic processes, the Stirling cycle replaces the adiabatic legs with **constant‑volume regenerative heat transfer**. This produces a conceptually elegant four‑step cycle:

- Two **isothermal** processes  
- Two **isochoric** (constant‑volume) regenerative processes  

Consider the following:

- $N$ moles of an ideal gas in a sealed cylinder–piston system  
- Expansion and compression occur reversibly  
- The cylinder head can be placed in contact with a hot reservoir at $T_H$ or a cold reservoir at $T_C$  
- A regenerator stores and releases heat during the constant‑volume steps  
- The shaft is connected to a motor from which we wish to extract work  

Based on these facts, we construct a **4‑step Stirling cycle** that produces net work.



### Pathways of the Stirling Cycle

| Step | Description | Pathway | Heat | Work |
|:---:|:-----------:|:--------|:-----|:-----|
| **1–2** | Regenerator (cold → hot) | Isochoric heating | $\dot{Q}_{\text{regen,in}}$ | $0$ |
| **2–3** | Hot reservoir | Isothermal expansion | $\dot{Q}_H$ | $-PdV$ |
| **3–4** | Regenerator (hot → cold) | Isochoric cooling | $\dot{Q}_{\text{regen,out}}$ | $0$ |
| **4–1** | Cold reservoir | Isothermal compression | $\dot{Q}_C$ | $-PdV$ |



### State Table for an Ideal‑Gas Stirling Cycle

| Step | Pathway | $T$ | $P$ | $\underline{V}$ | $\underline{S}$ | $\underline{U}$ |
|:---:|:--------|:-----:|:------:|:------------------:|:------------------:|:------------------:|
| **1** | Cold isotherm | $T_C$ | $P_1$ | $\underline{V}_1$ | $0$ | $0$ |
| **1–2** | Isochoric heating | | | | | |
| **2** | | $T_H$ | $P_2 = P_1(T_H/T_C)$ | $\underline{V}_1$ | $C_v^{IG}\ln(T_H/T_C)$ | $C_v^{IG}(T_H - T_C)$ |
| **2–3** | Isothermal expansion | | | | | |
| **3** | | $T_H$ | $P_3 = P_2/K$ | $\underline{V}_3 = K\underline{V}_1$ | $C_v^{IG}\ln(T_H/T_C) + R\ln K$ | $C_v^{IG}(T_H - T_C)$ |
| **3–4** | Isochoric cooling | | | | | |
| **4** | | $T_C$ | $P_4 = P_3(T_C/T_H)$ | $\underline{V}_3$ | $R\ln K$ | $0$ |
| **4–1** | Isothermal compression | | | | | |


### Specified Quantities

- Hot‑reservoir temperature: $T_H$  
- Cold‑reservoir temperature: $T_C$  
- Expansion ratio: $K = \underline{V}_3 / \underline{V}_1$  
- Initial pressure: $P_1$  
- Reference state: $T_R = T_C$, $\underline{S}_1 = 0$  
- Ideal‑gas heat capacity: $C_v^{IG}$ (constant)  



### Governing Relations Used

#### Equation of State
$$
P\underline{V} = RT
$$

#### Entropy Change of an Ideal Gas
$$
dS = \frac{C_v^{IG}}{T}dT + \frac{R}{\underline{V}}d\underline{V}
$$

#### Isothermal Steps
$$
\Delta U = 0,\qquad Q = W = RT\ln\left(\frac{V_f}{V_i}\right)
$$

#### Isochoric Steps
$$
W = 0,\qquad Q = \Delta U = C_v^{IG}(T_f - T_i)
$$



### Efficiency of the Ideal Stirling Cycle (Perfect Regenerator)

Using the pathway definitions:

- **1–2:** Isochoric heating (regenerator)  
- **2–3:** Isothermal expansion at $T_H$  
- **3–4:** Isochoric cooling (regenerator)  
- **4–1:** Isothermal compression at $T_C$  

#### 1–2: Isochoric heating
$$
\Delta U_{1\to2} = C_v^{IG}(T_H - T_C),\qquad Q_{1\to2} = \Delta U_{1\to2}
$$
Internal heat only (regenerator).

#### 2–3: Isothermal expansion at $T_H$
$$
W_{2\to3} = RT_H \ln K,\qquad Q_{2\to3} = RT_H \ln K
$$

#### 3–4: Isochoric cooling
$$
Q_{3\to4} = C_v^{IG}(T_C - T_H)
$$
Stored in regenerator.

#### 4–1: Isothermal compression at $T_C$
$$
W_{4\to1} = -RT_C \ln K,\qquad Q_{4\to1} = -RT_C \ln K
$$

#### External heat terms
$$
Q_{in} = RT_H \ln K,\qquad Q_{out} = -RT_C \ln K
$$

#### Net work
$$
W_{cycle} = R(T_H - T_C)\ln K
$$

#### Efficiency
$$
\eta = 1 + \frac{Q_{out}}{Q_{in}}
= 1 - \frac{RT_C\ln K}{RT_H\ln K}
= 1 - \frac{T_C}{T_H}
$$


### Final Result
$$
\boxed{\eta_{\text{Stirling}} = 1 - \frac{T_C}{T_H}}
$$

Even though the **paths** in $p\!-\!V$ and $T\!-\!S$ space differ from the Carnot cycle, a perfectly regenerated Stirling cycle can, in principle, achieve **Carnot efficiency**.


```python
import numpy as np
import matplotlib.pyplot as plt

R = 8.314  # J/mol/K

def Stirling_IG(Th=800, Tc=298, P1=1, Cv=28, K=10):
    """
    Ideal-gas Stirling cycle.

    Convention:
      1→2 : Isochoric heating (Tc → Th)
      2→3 : Isothermal expansion at Th
      3→4 : Isochoric cooling (Th → Tc)
      4→1 : Isothermal compression at Tc

    Inputs:
        Th : hot reservoir temperature (K)
        Tc : cold reservoir temperature (K)
        P1 : minimum pressure (bar) at state 1 (cold, small V)
        Cv : constant-volume heat capacity (J/mol/K)
        K  : expansion ratio (V3/V2 = V4/V1)

    Outputs:
        V_, P_, T_, S_, U_, A_ : state-point arrays
        PVpath_, TSpath_       : pathway arrays for plotting
    """

    Cp = Cv + R

    # -
    # STATE POINTS
    # -
    # State 1: cold, low volume
    P1 = P1 * 1e5          # bar → Pa
    T1 = Tc
    V1 = R * T1 / P1
    S1 = 0.0
    U1 = 0.0
    A1 = U1 - T1 * S1

    # State 2: isochoric heating to Th (V2 = V1)
    V2 = V1
    T2 = Th
    P2 = R * T2 / V2
    S2 = S1 + Cv * np.log(T2 / T1)
    U2 = Cv * (T2 - T1)
    A2 = U2 - T2 * S2

    # State 3: isothermal expansion at Th (V3 = K * V2)
    V3 = K * V2
    T3 = Th
    P3 = R * T3 / V3
    S3 = S2 + R * np.log(V3 / V2)
    U3 = U2  # isothermal ideal gas: U depends only on T
    A3 = U3 - T3 * S3

    # State 4: isochoric cooling back to Tc (V4 = V3)
    V4 = V3
    T4 = Tc
    P4 = R * T4 / V4
    S4 = S3 + Cv * np.log(T4 / T3)
    U4 = Cv * (T4 - T1)  # should return to 0
    A4 = U4 - T4 * S4

    # Check closure (numerically small drift is okay)
    # V1 == V1, V2 == V1, V3 == V4, etc.

    # -
    # STATE ARRAYS (converted)
    # -
    V_ = np.array([V1, V2, V3, V4, V1]) * 1e6  # m³/mol → cm³/mol
    P_ = np.array([P1, P2, P3, P4, P1]) / 1e5  # Pa → bar
    T_ = np.array([T1, T2, T3, T4, T1]) - 273.15
    S_ = np.array([S1, S2, S3, S4, S1])
    U_ = np.array([U1, U2, U3, U4, U1])
    A_ = np.array([A1, A2, A3, A4, A1])

    # -
    # PV PATHWAYS
    # -
    # 1→2: isochoric heating (vertical line in p–V)
    T12 = np.linspace(T1, T2)
    V12 = np.full_like(T12, V1)
    P12 = R * T12 / V12

    # 2→3: isothermal expansion at Th
    V23 = np.logspace(np.log10(V2), np.log10(V3))
    P23 = R * Th / V23

    # 3→4: isochoric cooling
    T34 = np.linspace(T3, T4)
    V34 = np.full_like(T34, V3)
    P34 = R * T34 / V34

    # 4→1: isothermal compression at Tc
    V41 = np.logspace(np.log10(V4), np.log10(V1))
    P41 = R * Tc / V41

    PVpath_ = [
        [V12 * 1e6, P12 / 1e5],
        [V23 * 1e6, P23 / 1e5],
        [V34 * 1e6, P34 / 1e5],
        [V41 * 1e6, P41 / 1e5],
    ]

    # -
    # TS PATHWAYS
    # -
    # 1→2: isochoric heating (S changes via Cv ln(T))
    T12 = np.linspace(T1, T2)
    S12 = S1 + Cv * np.log(T12 / T1)

    # 2→3: isothermal expansion at Th (T constant, S changes via R ln(V))
    S23 = np.linspace(S2, S3)
    T23 = np.full_like(S23, Th)

    # 3→4: isochoric cooling
    T34 = np.linspace(T3, T4)
    S34 = S3 + Cv * np.log(T34 / T3)

    # 4→1: isothermal compression at Tc
    S41 = np.linspace(S4, S1)
    T41 = np.full_like(S41, Tc)

    TSpath_ = [
        [T12 - 273.15, S12],
        [T23 - 273.15, S23],
        [T34 - 273.15, S34],
        [T41 - 273.15, S41],
    ]

    return V_, P_, T_, S_, U_, A_, PVpath_, TSpath_

V_,P_,T_,S_,U_,A_,PVpath_,TSpath_ = Stirling_IG(Th=900, Tc=300, Cv=12, K=10)

labels = ["1", "2", "3", "4", "1"]

fig, ax = plt.subplots(dpi=300, figsize=(5,2), ncols=2)
fig.subplots_adjust(wspace=0.5)

colors = ['r','r','b','b']

# -
# PV plot with labels
# -
ax[0].plot(V_, P_, marker='s', linestyle='')

for i, path in enumerate(PVpath_):
    ax[0].plot(path[0], path[1], c=colors[i])

# Add labels near each state point
for i in range(4):
    ax[0].text(V_[i]*1.05, P_[i]*1.05, labels[i], fontsize=8)

ax[0].set_xscale('log')
ax[0].set_yscale('log')
ax[0].set_xlabel(r'$V$ (cm$^3$/mol)')
ax[0].set_ylabel(r'$P$ (bar)')


# -
# TS plot with labels
# -
ax[1].plot(S_, T_, marker='s', linestyle='')

for i, path in enumerate(TSpath_):
    ax[1].plot(path[1], path[0], c=colors[i])

# Add labels near each state point
for i in range(4):
    ax[1].text(S_[i]*1.02, T_[i]*1.02, labels[i], fontsize=8)

ax[1].set_xlabel(r'$S$ (J/mol/K)')
ax[1].set_ylabel(r'$T$ (°C)')
```

## Ericsson Cycle

The Ericsson cycle is a regenerative, closed‑cycle heat engine that uses an ideal gas as its working fluid. It is closely related to the Brayton cycle, but replaces the adiabatic legs with **isothermal** processes and the constant‑pressure legs with **constant‑pressure regenerative heat transfer**.  

This produces a conceptually elegant four‑step cycle:

- Two **isothermal** processes  
- Two **isobaric** (constant‑pressure) regenerative processes  

Consider the following:

- $N$ moles of an ideal gas in a piston–cylinder or steady‑flow device  
- Expansion and compression occur reversibly  
- The working fluid exchanges heat with a hot reservoir at $T_H$ and a cold reservoir at $T_C$  
- A regenerator stores and releases heat during the constant‑pressure steps  
- The shaft is connected to a motor from which we extract work  

### Pathways of the Ericsson Cycle

| Step | Description | Pathway | Heat | Work |
|:---:|:-----------:|:--------|:-----|:-----|
| **1–2** | Regenerator (cold → hot) | Isobaric heating | $\dot{Q}_{\text{regen,in}}$ | $-P\,dV$ |
| **2–3** | Hot reservoir | Isothermal expansion | $\dot{Q}_H$ | $-P\,dV$ |
| **3–4** | Regenerator (hot → cold) | Isobaric cooling | $\dot{Q}_{\text{regen,out}}$ | $-P\,dV$ |
| **4–1** | Cold reservoir | Isothermal compression | $\dot{Q}_C$ | $-P\,dV$ |

### State Table for an Ideal‑Gas Ericsson Cycle

Pressure ratio:
$$
r_p = \frac{P_2}{P_1}
$$

| Step | Pathway | $T$ | $P$ | $\underline{V}$ | $\underline{S}$ | $\underline{U}$ |
|:---:|:--------|:-----:|:------:|:------------------:|:------------------:|:------------------:|
| **1** | Cold isotherm | $T_C$ | $P_1$ | $\underline{V}_1$ | $0$ | $0$ |
| **1–2** | Isobaric heating | | | | | |
| **2** | | $T_H$ | $P_1$ | $\underline{V}_2 = \underline{V}_1 \frac{T_H}{T_C}$ | $C_p^{IG}\ln\!\left(\frac{T_H}{T_C}\right)$ | $C_v^{IG}(T_H - T_C)$ |
| **2–3** | Isothermal expansion | | | | | |
| **3** | | $T_H$ | $P_3 = \frac{P_1}{r_p}$ | $\underline{V}_3 = r_p\,\underline{V}_2$ | $C_p^{IG}\ln\!\left(\frac{T_H}{T_C}\right) + R\ln r_p$ | $C_v^{IG}(T_H - T_C)$ |
| **3–4** | Isobaric cooling | | | | | |
| **4** | | $T_C$ | $P_3$ | $\underline{V}_4 = \underline{V}_3 \frac{T_C}{T_H}$ | $R\ln r_p$ | $0$ |
| **4–1** | Isothermal compression | | | | | |


### Governing Relations

#### Equation of State
$$
P\underline{V} = RT
$$

#### Entropy Change of an Ideal Gas
$$
dS = \frac{C_v^{IG}}{T}\,dT + \frac{R}{\underline{V}}\,d\underline{V}
$$
$$
dS = \frac{C_p^{IG}}{T}\,dT - \frac{R}{P}\,dP
$$

#### Isothermal Steps
$$
\Delta U = 0,\qquad Q = W = RT\ln\!\left(\frac{V_f}{V_i}\right)
$$

#### Isobaric Steps
$$
W = P(\underline{V}_f - \underline{V}_i),\qquad 
Q = \Delta H = C_p^{IG}(T_f - T_i)
$$


### Efficiency of the Ideal Ericsson Cycle (Perfect Regenerator)

#### Heat in (isothermal at $T_H$)
$$
Q_{\text{in}} = RT_H \ln r_p
$$

### Heat out (isothermal at $T_C$)
$$
Q_{\text{out}} = RT_C \ln r_p
$$

#### Net work
$$
W_{\text{cycle}} = R(T_H - T_C)\ln r_p
$$

#### Efficiency
$$
\eta = 1 - \frac{Q_{\text{out}}}{Q_{\text{in}}}
      = 1 - \frac{T_C}{T_H}
$$

$$
\boxed{\eta_{\text{Ericsson}} = 1 - \frac{T_C}{T_H}}
$$



```python
import numpy as np
import matplotlib.pyplot as plt

R = 8.314  # J/mol/K

def Ericsson_IG(Th=800, Tc=300, P1=1, rp=5, Cv=20):
    """
    Ideal-gas Ericsson cycle.

    Convention:
      1→2 : Isobaric heating (regenerator + hot side)
      2→3 : Isothermal expansion at Th
      3→4 : Isobaric cooling (regenerator + cold side)
      4→1 : Isothermal compression at Tc

    Inputs:
        Th : hot reservoir temperature (K)
        Tc : cold reservoir temperature (K)
        P1 : initial pressure (bar)
        rp : pressure ratio (P2/P3 = rp)
        Cv : constant-volume heat capacity (J/mol/K)

    Outputs:
        V_, P_, T_, S_, U_, A_ : state-point arrays
        PVpath_, TSpath_       : pathway arrays for plotting
    """

    Cp = Cv + R

    # -
    # STATE POINTS
    # -
    P1 = P1 * 1e5  # bar → Pa
    T1 = Tc
    V1 = R * T1 / P1
    S1 = 0
    U1 = Cv * T1
    A1 = U1 - T1 * S1

    # State 2: isobaric heating to Th
    P2 = P1
    T2 = Th
    V2 = R * T2 / P2
    S2 = S1 + Cp * np.log(T2 / T1)
    U2 = Cv * T2
    A2 = U2 - T2 * S2

    # State 3: isothermal expansion at Th
    P3 = P2 / rp
    T3 = Th
    V3 = R * T3 / P3
    S3 = S2 + R * np.log(V3 / V2)
    U3 = Cv * T3
    A3 = U3 - T3 * S3

    # State 4: isobaric cooling back to Tc
    P4 = P3
    T4 = Tc
    V4 = R * T4 / P4
    S4 = S3 + Cp * np.log(T4 / T3)
    U4 = Cv * T4
    A4 = U4 - T4 * S4

    # -
    # STATE ARRAYS (converted)
    # -
    V_ = np.array([V1, V2, V3, V4, V1]) * 1e6  # m³/mol → cm³/mol
    P_ = np.array([P1, P2, P3, P4, P1]) / 1e5  # Pa → bar
    T_ = np.array([T1, T2, T3, T4, T1]) - 273.15
    S_ = np.array([S1, S2, S3, S4, S1])
    U_ = np.array([U1, U2, U3, U4, U1])
    A_ = np.array([A1, A2, A3, A4, A1])

    # -
    # PV PATHWAYS
    # -
    # 1→2: isobaric heating
    T12 = np.linspace(T1, T2)
    V12 = R * T12 / P1
    P12 = np.full_like(T12, P1)

    # 2→3: isothermal expansion
    V23 = np.logspace(np.log10(V2), np.log10(V3))
    P23 = R * Th / V23

    # 3→4: isobaric cooling
    T34 = np.linspace(T3, T4)
    V34 = R * T34 / P3
    P34 = np.full_like(T34, P3)

    # 4→1: isothermal compression
    V41 = np.logspace(np.log10(V4), np.log10(V1))
    P41 = R * Tc / V41

    PVpath_ = [
        [V12 * 1e6, P12 / 1e5],
        [V23 * 1e6, P23 / 1e5],
        [V34 * 1e6, P34 / 1e5],
        [V41 * 1e6, P41 / 1e5],
    ]

    # -
    # TS PATHWAYS
    # -
    # 1→2: isobaric heating
    T12 = np.linspace(T1, T2)
    S12 = S1 + Cp * np.log(T12 / T1)

    # 2→3: isothermal expansion
    S23 = np.linspace(S2, S3)
    T23 = np.full_like(S23, Th)

    # 3→4: isobaric cooling
    T34 = np.linspace(T3, T4)
    S34 = S3 + Cp * np.log(T34 / T3)

    # 4→1: isothermal compression
    S41 = np.linspace(S4, S1)
    T41 = np.full_like(S41, Tc)

    TSpath_ = [
        [T12 - 273.15, S12],
        [T23 - 273.15, S23],
        [T34 - 273.15, S34],
        [T41 - 273.15, S41],
    ]

    return V_, P_, T_, S_, U_, A_, PVpath_, TSpath_
```

```python
V_,P_,T_,S_,U_,A_,PVpath_,TSpath_ = Ericsson_IG(Th=900, Tc=300, P1=1, rp=10, Cv=20)

labels = ["1", "2", "3", "4", "1"]
colors = ['r','r','b','b']

fig, ax = plt.subplots(dpi=300, figsize=(5,2), ncols=2)
fig.subplots_adjust(wspace=0.5)

# -
# PV plot
# -
ax[0].plot(V_, P_, marker='s', linestyle='')
for i, path in enumerate(PVpath_):
    ax[0].plot(path[0], path[1], c=colors[i])

for i in range(4):
    ax[0].text(V_[i]*1.05, P_[i]*1.05, labels[i], fontsize=8)

ax[0].set_xscale('log')
ax[0].set_yscale('log')
ax[0].set_xlabel(r'$V$ (cm$^3$/mol)')
ax[0].set_ylabel(r'$P$ (bar)')


# -
# TS plot
# -
ax[1].plot(S_, T_, marker='s', linestyle='')
for i, path in enumerate(TSpath_):
    ax[1].plot(path[1], path[0], c=colors[i])

for i in range(4):
    ax[1].text(S_[i]*1.02, T_[i]*1.02, labels[i], fontsize=8)

ax[1].set_xlabel(r'$S$ (J/mol/K)')
ax[1].set_ylabel(r'$T$ (°C)')
```

## Otto Cycle

### **Pathways of the Otto Cycle**

| Step | Description | Pathway | Heat | Work |
|:---:|:-----------:|:--------|:-----|:-----|
| **1–2** | Compression stroke | Isentropic compression | $0$ | $-P\,dV$ |
| **2–3** | Combustion | Isochoric heat addition | $\dot{Q}_H$ | $0$ |
| **3–4** | Power stroke | Isentropic expansion | $0$ | $-P\,dV$ |
| **4–1** | Exhaust / cooling | Isochoric heat rejection | $\dot{Q}_C$ | $0$ |


### **State Table for an Ideal‑Gas Otto Cycle**

Compression ratio:
$$
r = \frac{\underline{V}_1}{\underline{V}_2}
$$

| Step | Pathway | $T$ | $P$ | $\underline{V}$ | $\underline{S}$ | $\underline{U}$ |
|:---:|:--------|:-----:|:------:|:------------------:|:------------------:|:------------------:|
| **1** | Start of compression | $T_1$ | $P_1$ | $\underline{V}_1$ | $0$ | $C_v T_1$ |
| **1–2** | Isentropic compression | | | | | |
| **2** | | $T_2 = T_1 r^{\gamma - 1}$ | $P_2 = P_1 r^\gamma$ | $\underline{V}_2 = \underline{V}_1/r$ | $0$ | $C_v T_2$ |
| **2–3** | Isochoric heat addition | | | | | |
| **3** | | $T_3$ | $P_3 = P_2 \left(\frac{T_3}{T_2}\right)$ | $\underline{V}_2$ | $C_v \ln\!\left(\frac{T_3}{T_2}\right)$ | $C_v T_3$ |
| **3–4** | Isentropic expansion | | | | | |
| **4** | | $T_4 = T_3 r^{1-\gamma}$ | $P_4 = P_3 r^{-\gamma}$ | $\underline{V}_1$ | $C_v \ln\!\left(\frac{T_3}{T_2}\right)$ | $C_v T_4$ |
| **4–1** | Isochoric heat rejection | | | | | |

Here $\gamma = C_p/C_v$.


### **Specified Quantities**

- Initial temperature: $T_1$  
- Initial pressure: $P_1$  
- Compression ratio: $r = V_1/V_2$  
- Heat‑capacity ratio: $\gamma = C_p/C_v$  
- Heat added: $Q_{\text{in}} = C_v (T_3 - T_2)$  
- Heat rejected: $Q_{\text{out}} = C_v (T_4 - T_1)$



### **Governing Relations**

#### **Equation of State**
$$
P\underline{V} = RT
$$

#### **Isentropic Relations**
$$
T V^{\gamma - 1} = \text{const}
$$
$$
P V^\gamma = \text{const}
$$

#### **Isochoric Heat Transfer**
$$
Q = \Delta U = C_v (T_f - T_i)
$$


### **Efficiency of the Ideal Otto Cycle**

$$
\eta = 1 - \frac{Q_C}{Q_H}
$$

#### Heat added
$$
Q_H = C_v (T_3 - T_2)
$$

#### Heat rejected
$$
Q_C = C_v (T_4 - T_1)
$$

Using isentropic relations:
$$
T_2 = T_1 r^{\gamma - 1}, \qquad
T_4 = T_3 r^{1 - \gamma}
$$

Substitute:
$$
\eta = 1 - \frac{T_4 - T_1}{T_3 - T_2}
$$

After algebra:
$$ \boxed{\eta_{\text{Otto}} = 1 - \frac{1}{r^{\gamma - 1}}} $$

This is the classic Otto‑cycle efficiency:  
**higher compression ratio → higher efficiency**, and **larger $\gamma$** also improves efficiency.


```python
import numpy as np
import matplotlib.pyplot as plt

R = 8.314  # J/mol/K

def Otto_IG(T1=300, P1=1, r=10, Cv=20):
    """
    Ideal-gas Otto cycle.

    Convention:
      1→2 : Isentropic compression
      2→3 : Isochoric heat addition
      3→4 : Isentropic expansion
      4→1 : Isochoric heat rejection

    Inputs:
        T1 : initial temperature (K)
        P1 : initial pressure (bar)
        r  : compression ratio (V1/V2)
        Cv : constant-volume heat capacity (J/mol/K)

    Outputs:
        V_, P_, T_, S_, U_, A_ : state-point arrays
        PVpath_, TSpath_       : pathway arrays for plotting
    """

    Cp = Cv + R
    gamma = Cp / Cv

    # -
    # STATE POINTS
    # -
    P1 = P1 * 1e5  # bar → Pa
    V1 = R * T1 / P1
    S1 = 0
    U1 = Cv * T1
    A1 = U1 - T1 * S1

    # State 2: isentropic compression
    V2 = V1 / r
    T2 = T1 * r**(gamma - 1)
    P2 = R * T2 / V2
    S2 = S1
    U2 = Cv * T2
    A2 = U2 - T2 * S2

    # State 3: isochoric heat addition
    # Choose a heat addition factor (e.g., T3 = a * T2)
    a = 2.5
    T3 = a * T2
    V3 = V2
    P3 = R * T3 / V3
    S3 = S2 + Cv * np.log(T3 / T2)
    U3 = Cv * T3
    A3 = U3 - T3 * S3

    # State 4: isentropic expansion
    V4 = V1
    T4 = T3 * (1/r)**(gamma - 1)
    P4 = R * T4 / V4
    S4 = S3
    U4 = Cv * T4
    A4 = U4 - T4 * S4

    # -
    # STATE ARRAYS (converted)
    # -
    V_ = np.array([V1, V2, V3, V4, V1]) * 1e6  # m³/mol → cm³/mol
    P_ = np.array([P1, P2, P3, P4, P1]) / 1e5  # Pa → bar
    T_ = np.array([T1, T2, T3, T4, T1]) - 273.15
    S_ = np.array([S1, S2, S3, S4, S1])
    U_ = np.array([U1, U2, U3, U4, U1])
    A_ = np.array([A1, A2, A3, A4, A1])

    # -
    # PV PATHWAYS
    # -
    # 1→2: isentropic compression
    V12 = np.logspace(np.log10(V1), np.log10(V2))
    P12 = P1 * (V1 / V12)**gamma

    # 2→3: isochoric heat addition
    T23 = np.linspace(T2, T3)
    V23 = np.full_like(T23, V2)
    P23 = R * T23 / V23

    # 3→4: isentropic expansion
    V34 = np.logspace(np.log10(V3), np.log10(V4))
    P34 = P3 * (V3 / V34)**gamma

    # 4→1: isochoric heat rejection
    T41 = np.linspace(T4, T1)
    V41 = np.full_like(T41, V1)
    P41 = R * T41 / V41

    PVpath_ = [
        [V12 * 1e6, P12 / 1e5],
        [V23 * 1e6, P23 / 1e5],
        [V34 * 1e6, P34 / 1e5],
        [V41 * 1e6, P41 / 1e5],
    ]

    # -
    # TS PATHWAYS
    # -
    # 1→2: isentropic (vertical line)
    T12 = np.logspace(np.log10(T1), np.log10(T2))
    S12 = np.full_like(T12, S1)

    # 2→3: isochoric heat addition
    T23 = np.linspace(T2, T3)
    S23 = S2 + Cv * np.log(T23 / T2)

    # 3→4: isentropic
    T34 = np.logspace(np.log10(T3), np.log10(T4))
    S34 = np.full_like(T34, S3)

    # 4→1: isochoric heat rejection
    T41 = np.linspace(T4, T1)
    S41 = S4 + Cv * np.log(T41 / T4)

    TSpath_ = [
        [T12 - 273.15, S12],
        [T23 - 273.15, S23],
        [T34 - 273.15, S34],
        [T41 - 273.15, S41],
    ]

    return V_, P_, T_, S_, U_, A_, PVpath_, TSpath_
```

```python
V_,P_,T_,S_,U_,A_,PVpath_,TSpath_ = Otto_IG(T1=300, P1=1, r=10, Cv=20)

labels = ["1", "2", "3", "4", "1"]
colors = ['k','r','k','b']

fig, ax = plt.subplots(dpi=300, figsize=(5,2), ncols=2)
fig.subplots_adjust(wspace=0.5)

# -
# PV plot
# -
ax[0].plot(V_, P_, marker='s', linestyle='')
for i, path in enumerate(PVpath_):
    ax[0].plot(path[0], path[1], c=colors[i])

# Labels
for i in range(4):
    ax[0].text(V_[i]*1.05, P_[i]*1.05, labels[i], fontsize=8)

ax[0].set_xscale('log')
ax[0].set_yscale('log')
ax[0].set_xlabel(r'$V$ (cm$^3$/mol)')
ax[0].set_ylabel(r'$P$ (bar)')


# -
# TS plot
# -
ax[1].plot(S_, T_, marker='s', linestyle='')
for i, path in enumerate(TSpath_):
    ax[1].plot(path[1], path[0], c=colors[i])

# Labels
for i in range(4):
    ax[1].text(S_[i]*1.02, T_[i]*1.02, labels[i], fontsize=8)

ax[1].set_xlabel(r'$S$ (J/mol/K)')
ax[1].set_ylabel(r'$T$ (°C)')
```

## Summary of Major Thermodynamic Cycles and Their Applications

### 1. Carnot Cycle
**Processes:**  
- 2 isothermal (heat in/out)  
- 2 adiabatic (isentropic)  

**Key idea:**  
Theoretical upper limit on efficiency for any heat engine. No real engine achieves it, but it defines the benchmark.

**Efficiency:**  
$$
\eta_{Carnot} = 1 - \frac{T_C}{T_H}
$$

**Real‑world applications:**  
- Not used directly in engines  
- **Benchmark** for evaluating real cycles  
- Foundation for **thermodynamic efficiency limits** in power plants, refrigerators, and heat pumps  
- Conceptual tool in engineering education  



### 2. Stirling Cycle
**Processes:**  
- 2 isothermal  
- 2 isochoric (regenerative)  

**Key idea:**  
Uses a **regenerator** to store/reuse heat, allowing Carnot‑level efficiency in principle.

**Efficiency (ideal):**  
$$
\eta_{Stirling} = 1 - \frac{T_C}{T_H}
$$

**Real‑world applications:**  
- **Stirling engines** for quiet, low‑maintenance power generation  
- **Cryocoolers** (Stirling refrigerators)  
- Submarines and spacecraft auxiliary power (high reliability, low noise)  
- Solar‑thermal engines (dish‑Stirling systems)  



### 3. Ericsson Cycle
**Processes:**  
- 2 isothermal  
- 2 isobaric (regenerative)  

**Key idea:**  
A regenerative version of the Brayton cycle with isothermal compression/expansion.  
Like Stirling, it can reach Carnot efficiency with perfect regeneration.

**Efficiency (ideal):**  
$$
\eta_{Ericsson} = 1 - \frac{T_C}{T_H}
$$

**Real‑world applications:**  
- Conceptual model for **regenerative gas turbines**  
- Basis for advanced **external‑combustion** and **closed‑cycle** engines  
- Educational model for understanding regeneration in Brayton‑type systems  



### 4. Otto Cycle
**Processes:**  
- 2 isentropic (compression/expansion)  
- 2 isochoric (heat addition/rejection)  

**Key idea:**  
Idealized model of spark‑ignition internal combustion engines.

**Efficiency:**  
$$
\eta_{Otto} = 1 - \frac{1}{r^{\gamma - 1}}
$$

**Real‑world applications:**  
- **Gasoline engines** (cars, motorcycles, small aircraft)  
- **Small generators**  
- Any spark‑ignition engine using premixed fuel–air combustion  




### Comparison of Major Thermodynamic Cycles

| Cycle | Process Sequence | Key Assumptions | Ideal Efficiency | Real‑World Applications | Distinguishing Features |
|------|------------------|-----------------|------------------|-------------------------|--------------------------|
| **Carnot** | 1. Isothermal expansion  2. Adiabatic expansion  3. Isothermal compression  4. Adiabatic compression | Reversible, ideal gas, perfect insulation, perfect heat transfer | $\eta = 1 - \frac{T_C}{T_H}$ | Benchmark for all heat engines; theoretical limit used in power plant analysis | Highest possible efficiency for given $T_H, T_C$; not realizable in practice |
| **Stirling** | 1. Isochoric heating (regen)  2. Isothermal expansion  3. Isochoric cooling (regen)  4. Isothermal compression | Perfect regenerator; ideal gas; reversible | $\eta = 1 - \frac{T_C}{T_H}$ | Stirling engines, cryocoolers, solar‑thermal engines, submarines | Uses a **regenerator**; can reach Carnot efficiency; very quiet and smooth |
| **Ericsson** | 1. Isobaric heating (regen)  2. Isothermal expansion  3. Isobaric cooling (regen)  4. Isothermal compression | Perfect regenerator; ideal gas; reversible | $\eta = 1 - \frac{T_C}{T_H}$ | Conceptual model for regenerative gas turbines; advanced external‑combustion engines | Brayton‑like but with isothermal heat transfer; also reaches Carnot efficiency |
| **Otto** | 1. Isentropic compression  2. Isochoric heat addition  3. Isentropic expansion  4. Isochoric heat rejection | Ideal gas; instantaneous combustion; reversible compression/expansion | $\eta = 1 - \frac{1}{r^{\gamma - 1}}$ | Gasoline engines, small generators, spark‑ignition engines | Efficiency depends on **compression ratio**; models spark‑ignition combustion |
| **Diesel** *(optional addition)* | 1. Isentropic compression  2. Isobaric heat addition  3. Isentropic expansion  4. Isochoric heat rejection | Ideal gas; constant‑pressure combustion | $\eta = 1 - \frac{1}{r^{\gamma - 1}} \cdot \frac{\rho^\gamma - 1}{\gamma(\rho - 1)}$ | Diesel engines, heavy vehicles, ships | Heat addition at constant pressure; higher efficiency than Otto at same $r$ |
| **Brayton** *(optional addition)* | 1. Isentropic compression  2. Isobaric heat addition  3. Isentropic expansion  4. Isobaric heat rejection | Ideal gas; steady‑flow turbine/compressor | $\eta = 1 - \frac{1}{r_p^{(\gamma - 1)/\gamma}}$ | Jet engines, gas turbines, power plants | Continuous‑flow analog of Otto; efficiency depends on **pressure ratio** |

### How they relate 

- **Carnot** defines the theoretical limit.  
- **Stirling** and **Ericsson** achieve Carnot efficiency *in principle* through perfect regeneration and isothermal heat transfer.  
- **Otto** trades ideal efficiency for **practicality**, modeling real spark‑ignition engines where combustion is rapid and nearly constant‑volume.
